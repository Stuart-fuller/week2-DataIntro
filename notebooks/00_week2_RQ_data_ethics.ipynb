## Research Questions and Answers

For every answer, distinguish between **known evidence**, **reasonable interpretation**, and **unknown information to investigate**.

### 1. Where did the data come from?

**Answer:**

Known evidence:
The dataset comes from The Open University. It contains records from seven selected modules, including student information, assessment results, and activity in the virtual learning 
environment.

Reasonable interpretation: 
It gives us a way to study how online learning activity relates to academic results.

Unknown to investigate: 
How well these seven modules represent University students more broadly.

### 2. Why was it originally collected?

**Answer:**

Known evidence: 
The Open University records information such as assignment submissions and uses student data for monitoring, early warnings, and improving teaching.

Reasonable interpretation: 
Some of this information can help identify students who may need extra support.

Unknown to investigate: 
We cannot say that every record in this published dataset was originally collected specifically to predict failure in the next assignment.


### 3. Who is represented?

**Answer:**

Known evidence: Students enrolled in seven selected Open University modules are represented. The data includes recorded gender, age band, region, and other student characteristics.

Reasonable interpretation: 
The dataset includes students from a range of backgrounds.

Unknown to investigate: 
We would need to count each group before saying how fully different genders, age bands, or regions are represented.

### 4. Who might be missing?

**Answer:** 
Known evidence: 
The published dataset contains students from seven selected Open University modules taught in 2013 and 2014, rather than every student or module at the university. The researchers selected four STEM modules and three social sciences modules. To be eligible for selection, a module had to have more than 500 students, at least two presentations, available virtual learning environment data, and a significant number of students who failed. This means smaller modules and modules that did not meet the other selection criteria are absent from the dataset. It also contains no records from campus-based universities.

Reasonable interpretation: 
Students at campus based universities may study differently. People with limited internet access or circumstances that make online study difficult may also be underrepresented.

Unknown to investigate: 
The dataset alone cannot tell us how many people were unable to enrol or participate because of those barriers.


### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** 
Known evidence: The published dataset is anonyminised, but includes age band, gender, region, education level, declared disability status, and academic results. It also records assessment and online activity data.

Reasonable interpretation: 
Even without names, these details should be handled carefully because the analysis could produce sensitive conclusions about groups of students.

Unknown to investigate: 
Whether any combination of fields could identify an individual in the data we use.

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:**

Known evidence: 
The dataset records assessment results, final outcomes, and students’ interactions with the virtual learning environment. 

Reasonable interpretation: 
Analysing these records together could reveal whether patterns such as regular engagement, changes in activity, or the timing of assignment submissions are associated with better results or with withdrawal. Early results and activity might also help identify students who could benefit from support. These would be associations, however the data could not prove that clicking on more materials or submitting earlier causes better grades.

information to investigate: 
We would need to test which patterns are actually present and whether an early warning model makes reliable predictions across different modules

### 7. Who could benefit from your findings?

**Answer:**

Known evidence: The dataset includes information that can be used to study learning activity alongside assessment outcomes. 

Reasonable interpretation: 
Students might use the findings to reflect on how they engage with course materials, while lecturers and university support staff might use them to recognise students who could benefit from timely help. If that support is effective, it could help reduce failures or withdrawals. 

Unknown information to investigate: 
The dataset alone cannot establish that a particular study habit improves grades, or that using a prediction to offer support would actually reduce withdrawal rates.

### 8. Who could be harmed or misrepresented?

**Answer:**

Known evidence: 
The published dataset covers seven selected Open University modules and includes characteristics such as age band and declared disability status, as well as learning activity and results. 

Reasonable interpretation: 
A student could be unfairly labelled “likely to fail” if a model makes an incorrect prediction, especially if that label affects how staff treat them or what opportunities they receive. Low recorded online activity might also reflect work, caring responsibilities, disability, or studying offline rather than a lack of effort. Because these students studied through the Open University, findings may not apply in the same way to students at campus-based universities; we should therefore avoid presenting them as conclusions about all university students. 

Unknown information to investigate: 
We would need to check whether errors differ between groups and whether the selected modules adequately represent the wider student population. we also need to verify that it is no possible to identify individuals if there is rare combinations of features which can narrow down the ID to a single person. This would mean the data is not full anomonous.
