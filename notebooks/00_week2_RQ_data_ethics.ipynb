### 1. Where did the data come from?

**Answer:** The dataset came from Open University

### 2. Why was it originally collected?

**Answer:** It was originally collected to predict what students were at risk of failing in their next assignments and to offer support. 


### 3. Who is represented?

**Answer:** Students of all genders, age bands, and regions across UK and Ireland (it is an online university) are represented in this dataset.

### 4. Who might be missing?

**Answer:** Students without reliable access to a computer or internet may be missing. It may also not fully represent people whose disabilities or personal circumstances make online study difficult.

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** The data is anonamised, but it includes sensitive information such as age band, gender, region, education level, disability status, and academic outcomes.

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:** Our analysis could show links between student activity, assessment behavior, and final results. For example, it may show whether students with early assignment submission likely to fail or withdraw. We could also build a model to identify students who may need extra support, although this would show correlation rather than prove a cause.

### 7. Who could benefit from your findings?

**Answer:** Students could benefit from knowing what habits lead to better results. Lecturers and universities could use the findings  to locate students who may be struggling and reduce  overall withdrawal or failure rates.

### 8. Who could be harmed or misrepresented?

**Answer:** Students could be harmed if a model incorrectly labels them as likely to fail. The results may not represent all college students because the dataset only includes online learners, who tend to be older and may have different work or family commitments than students attending college in person. Therefore, we should avoid applying our conclusions to all students.
